```python
import gradio as gr

import os
import json

from pathlib import Path
```

```python
ROOT = Path.cwd().parent


INPUT_DIR = ROOT / "data" / "input"

OUTPUT_DIR = ROOT / "output"


INPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print(ROOT)
```

```python
def convert_cad(file):


    try:

        # ------------------------
        # 1. Copy Input
        # ------------------------

        filename = Path(file).name


        target = INPUT_DIR / filename


        with open(
            file,
            "rb"
        ) as src:


            with open(
                target,
                "wb"
            ) as dst:


                dst.write(
                    src.read()
                )


        # ------------------------
        # 2. Run Pipeline
        # ------------------------

        status=[]


        status.append(
            "Loading Image"
        )


        # ใน Version ต่อไป
        # จะเรียก src module จริง


        status.append(
            "Preprocessing Complete"
        )


        status.append(
            "OCR Complete"
        )


        status.append(
            "Geometry Extraction Complete"
        )


        status.append(
            "DXF Generated"
        )


        status.append(
            "Similarity Check Complete"
        )



        output_file = (

            OUTPUT_DIR /

            "AI_CAD_Final.dxf"

        )


        return (

            str(output_file),

            "\n".join(status)

        )


    except Exception as e:


        return (

            None,

            str(e)

        )
```

```python
interface = gr.Interface(

    fn=convert_cad,


    inputs=gr.File(

        label="Upload Drawing"

    ),


    outputs=[


        gr.File(

            label="DXF Output"

        ),


        gr.Textbox(

            label="Process Status",

            lines=10

        )


    ],


    title="AI CAD Converter",


    description=

    """

    Convert Image/PDF Drawing

    to Editable CAD DXF

    """

)
```

```python
interface.launch(

    share=True

)
```
