```python
import fitz
import json

from pathlib import Path

from PIL import Image
```

```python
# ======================================================
# Project Path
# ======================================================


ROOT = Path.cwd().parent


INPUT_DIR = (

ROOT /

"data" /

"input"

)


PROCESSED_DIR = (

ROOT /

"data" /

"processed"

)


PROCESSED_DIR.mkdir(

    exist_ok=True

)


print(ROOT)
```

```python
# ======================================================
# Find PDF
# ======================================================


pdf_files=list(

    INPUT_DIR.glob(

        "*.pdf"

    )

)


if len(pdf_files)==0:

    raise Exception(

        "No PDF found in data/input"

    )


PDF_FILE=pdf_files[0]


print(

"PDF:",

PDF_FILE

)
```

```python
# ======================================================
# Open PDF
# ======================================================


pdf = fitz.open(

    PDF_FILE

)


print(

"Pages:",

pdf.page_count

)
```

```python
# ======================================================
# Render Function
# ======================================================


def render_page(

    page,

    dpi=300

):


    zoom = dpi / 72


    matrix = fitz.Matrix(

        zoom,

        zoom

    )


    pix = page.get_pixmap(

        matrix=matrix

    )


    image = Image.frombytes(

        "RGB",

        [

            pix.width,

            pix.height

        ],

        pix.samples

    )


    return image
```

```python
# ======================================================
# PDF To Image
# ======================================================


PAGE_IMAGES=[]


for i,page in enumerate(pdf):


    img=render_page(

        page,

        dpi=300

    )


    output=(

        PROCESSED_DIR /

        f"page_{i+1}.png"

    )


    img.save(

        output

    )


    PAGE_IMAGES.append(

        str(output)

    )


    print(

        "Saved:",

        output

    )
```

```python
import matplotlib.pyplot as plt


img=Image.open(

    PAGE_IMAGES[0]

)


plt.figure(

    figsize=(10,10)

)


plt.imshow(img)


plt.axis("off")


plt.title(

"PDF Render"

)


plt.show()
```

```python
# ======================================================
# Resolution Check
# ======================================================


width,height=img.size


print(

"Width:",

width

)


print(

"Height:",

height

)


if width < 2000:

    print(

    "Warning: Low Resolution"

    )

else:

    print(

    "Resolution OK"

    )
```

```python
# ======================================================
# PDF State
# ======================================================


PDF_STATE={


"source_pdf":

str(PDF_FILE),


"pages":

len(PAGE_IMAGES),


"images":

PAGE_IMAGES,


"dpi":

300


}



PDF_STATE
```

```python
# ======================================================
# Save PDF State
# ======================================================


PDF_STATE_FILE=(

ROOT /

"config" /

"pdf_state.json"

)


with open(

    PDF_STATE_FILE,

    "w",

    encoding="utf-8"

) as f:


    json.dump(

        PDF_STATE,

        f,

        indent=4

    )



print(

"PDF State Saved"

)
```

```python
# ======================================================
# Pipeline Input Selector
# ======================================================


INPUT_MODE="PDF"


if INPUT_MODE=="PDF":


    CURRENT_INPUT = PAGE_IMAGES[0]


else:


    CURRENT_INPUT = None



print(

CURRENT_INPUT

)
```
