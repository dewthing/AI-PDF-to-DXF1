```python
import json

from pathlib import Path

import ezdxf
```

```python
ROOT = Path.cwd().parent


CAD_OBJECT_FILE = (

ROOT /

"temp" /

"cad_objects.json"

)


with open(

    CAD_OBJECT_FILE,

    encoding="utf-8"

) as f:


    CAD_OBJECTS=json.load(f)



print(

"Objects:",

len(CAD_OBJECTS)

)
```

```python
LAYER_MAP = {


"WALL":
"A-WALL",


"DOOR":
"A-DOOR",


"WINDOW":
"A-WINDOW",


"ROOM":
"A-ROOM",


"TEXT":
"A-ANNO-TEXT",


"CIRCLE":
"A-SYMBOL"


}


LAYER_MAP
```

```python
doc = ezdxf.new(

    "R2018"

)


msp = doc.modelspace()
```

```python
for name in LAYER_MAP.values():


    doc.layers.add(

        name

    )


print(

"Layers Created"

)
```

```python
for obj in CAD_OBJECTS:


    obj_type=obj["type"]


    layer=LAYER_MAP.get(

        obj_type,

        "0"

    )


    geo=obj["geometry"]



    if obj_type=="WALL":


        msp.add_line(

            geo["start"],

            geo["end"],

            dxfattribs={

                "layer":

                layer

            }

        )
```

```python
for obj in CAD_OBJECTS:


    if obj["type"]=="ROOM":


        pts=obj["geometry"]["points"]


        msp.add_lwpolyline(

            pts,

            close=True,

            dxfattribs={

                "layer":

                "A-ROOM"

            }

        )
```

```python
for obj in CAD_OBJECTS:


    layer=LAYER_MAP.get(

        obj["type"],

        "0"

    )


    print(

        obj["type"],

        "→",

        layer,

        "confidence:",

        obj["confidence"]

    )
```

```python
OUTPUT = (

ROOT /

"output" /

"AI_CAD_Intelligent.dxf"

)


doc.saveas(

    OUTPUT

)


print(

OUTPUT

)
```
